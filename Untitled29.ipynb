```python
# coding: utf-8
import wikipedia # wikipedia ライブラリを使用
import os
from google.colab import files
import time # リクエスト間の待機時間のため
# エラーハンドリングのために例外クラスをインポート
from wikipedia.exceptions import DisambiguationError, PageError
```

```python
# --- 設定 ---
WIKI_LANG = 'ja'  # Wikipediaの言語を日本語に設定
PAGE_TITLE_QUERY = 'ONE PIECE' # 検索するキーワード
OUTPUT_DIR = '/content/knowledge_base' # 保存先ディレクトリ (Colabのデフォルト)
OUTPUT_FILENAME = 'one_piece_wiki.txt' # 保存ファイル名

# --- ディレクトリ作成 ---
os.makedirs(OUTPUT_DIR, exist_ok=True)
output_path = os.path.join(OUTPUT_DIR, OUTPUT_FILENAME)

# --- Wikipedia ライブラリの設定 ---
wikipedia.set_lang(WIKI_LANG)
print(f"Wikipedia言語を '{WIKI_LANG}' に設定しました。")

# --- ページ取得 ---
print(f"'{PAGE_TITLE_QUERY}' に関連するWikipediaページを検索中...")

page_content = None
actual_page_title = None

try:
    # 1. キーワードで記事候補を検索
    search_results = wikipedia.search(PAGE_TITLE_QUERY)

    if not search_results:
        print(f"'{PAGE_TITLE_QUERY}' に一致する記事候補は見つかりませんでした。")
    else:
        # 2. 検索結果の最初の候補を選択
        selected_title = search_results[0]
        print(f"最も関連性の高い候補として '{selected_title}' を選択しました。")
        print(f"ページ '{selected_title}' の内容を取得します...")

        # 3. 選択した記事のページを取得
        # page() 関数は曖昧さ回避ページやページが見つからない場合に例外を発生させる可能性がある
        time.sleep(1) # 短時間での連続リクエストを避ける
        wiki_page = wikipedia.page(selected_title)
        page_content = wiki_page.content # ページの本文テキストを取得
        actual_page_title = wiki_page.title # 実際に取得されたページのタイトル
        print(f"ページ '{actual_page_title}' を取得しました。")

# エラーハンドリング
except DisambiguationError as e:
    # page() が曖昧さ回避ページに当たった場合 (search()で絞り込んでも発生する可能性はある)
    print(f"エラー: 選択された候補 '{selected_title}' は曖昧さ回避ページでした。候補: {e.options}")
    # ここで e.options からユーザーに選ばせるなどの処理も可能
except PageError:
    # page() で指定されたページが見つからなかった場合 (通常はsearch()で見つかるはずだが念のため)
    print(f"エラー: ページ '{selected_title}' が見つかりませんでした。")
except Exception as e:
    # その他の予期せぬエラー
    print(f"ページの取得中に予期せぬエラーが発生しました: {e}")


# --- ファイルへの保存とダウンロード ---
if page_content:
    try:
        with open(output_path, 'w', encoding='utf-8') as f:
            f.write(page_content)
        print(f"ページ '{actual_page_title}' の内容を '{output_path}' に保存しました。")
        print("-" * 20)
        # 内容の最初の500文字を表示 (確認用)
        print("保存された内容の冒頭:")
        print(page_content[:500] + "...")
        print("-" * 20)

        # --- ファイルのダウンロード ---
        print(f"'{output_path}' のダウンロードを開始します...")
        files.download(output_path)
        print("ダウンロードが完了しました（ブラウザのダウンロードフォルダを確認してください）。")

    except Exception as e:
        print(f"ファイル処理中にエラーが発生しました: {e}")
else:
    print("取得できるページ内容がなかったため、ファイルは保存・ダウンロードされませんでした。")

!pip install wikipedia
```
